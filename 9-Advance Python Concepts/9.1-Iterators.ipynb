### Iterators
Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.


```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
my_list = [1, 2, 3, 4, 5, 6]
iterator = iter(my_list)

print("=== The List ===")
print(f"List value: {my_list}")
print(f"List type: {type(my_list)}")
print(f"List id (decimal): {id(my_list)}")
print(f"List id (hex): {hex(id(my_list))}")

print("\n=== The Iterator ===")
print(f"Iterator: {iterator}")
print(f"Iterator type: {type(iterator)}")
print(f"Iterator id (decimal): {id(iterator)}")
print(f"Iterator id (hex): {hex(id(iterator))}")

print("\n=== Are They the Same Object? ===")
print(f"Same object? {id(my_list) == id(iterator)}")

next(iterator)
print("\n=== The Iterator ===")
print(f"Iterator: {iterator}")
print(f"Iterator type: {type(iterator)}")
print(f"Iterator id (decimal): {id(iterator)}")
print(f"Iterator id (hex): {hex(id(iterator))}")

dir(iterator)
```

```python
print("=== Eager Loading (List) ===")
print("Creating list...")
eager_list = [x**2 for x in range(5)]
print(f"List created: {eager_list}")
print("All values computed immediately!\n")

print("=== Lazy Loading (Generator/Iterator) ===")
print("Creating generator...")
lazy_gen = (x**2 for x in range(5))
print(f"Generator created: {lazy_gen}")
print("No values computed yet!\n")

print("Getting first value...")
print(f"Value: {next(lazy_gen)}")  # NOW 0² is computed
print("\nGetting second value...")
print(f"Value: {next(lazy_gen)}")  # NOW 1² is computed
print("\nGetting third value...")
print(f"Value: {next(lazy_gen)}")  # NOW 2² is computed
```

```python
# List comprehension - []
my_list = [x**2 for x in range(5)]
print(f"[]: {my_list}")
print(f"Type: {type(my_list)}\n")

# Generator expression - ()
my_gen = (x**2 for x in range(5))
print(f"(): {my_gen}")
print(f"Type: {type(my_gen)}\n")

# Set comprehension - {}
my_set = {x**2 for x in range(5)}
print(f"{{}}: {my_set}")
print(f"Type: {type(my_set)}\n")

# Dict comprehension - {key: value}
my_dict = {x: x**2 for x in range(5)}
print(f"{{k:v}}: {my_dict}")
print(f"Type: {type(my_dict)}")
```

```python
# Generator to list
gen = (x**2 for x in range(5))
lst = list(gen)  # Converts to [0, 1, 4, 9, 16]
print(lst)

# List to generator
lst = [1, 2, 3, 4, 5]
gen = (x for x in lst)  # Creates generator from list
print(next(gen))  # 1

# Or use iter()
gen = iter(lst)
print(next(gen))  # 1
```
