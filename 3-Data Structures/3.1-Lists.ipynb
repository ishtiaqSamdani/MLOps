##### List Comprehension

Basics Syantax            [expression for item in iterable]

with conditional logic    [expression for item in iterable if condition]

Nested List Comprehension [expression for item1 in iterable1 for item2 in iterable2]




```python
### Basic List Comphrension

sqaure=[num**2 for num in range(10)]
print(sqaure)
```

```python
### List Comprehension with Condition
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)

print(lst)
        
```

```python
even_numbers=[num for num in range(10) if num%2==0]
print(even_numbers)
```

```python
## Nested List Comphrension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]

print(pair)
```

```python
## List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)  # Output: [5, 5, 6, 4, 13]

```

#### Conclusion

## Removing Elements from Lists

There are several ways to remove elements from lists in Python:

### 1. Removing All Occurrences of an Element
### 2. Removing Elements by Index
### 3. Other Removal Methods


```python
### Method 1: Using remove() - Removes only the FIRST occurrence
# This only removes the first occurrence of the element

numbers = [1, 2, 3, 2, 4, 2, 5]
print("Original list:", numbers)

numbers.remove(2)  # Removes only the first occurrence of 2
print("After remove(2):", numbers)
```

```python
### Method 2: Using List Comprehension - Removes ALL occurrences
# This removes all occurrences of the element

numbers = [1, 2, 3, 2, 4, 2, 5]
print("Original list:", numbers)

# Remove all occurrences of 2
numbers = [x for x in numbers if x != 2]
print("After removing all 2s:", numbers)
```

```python
### Method 3: Using filter() - Removes ALL occurrences
# Another way to remove all occurrences

numbers = [1, 2, 3, 2, 4, 2, 5]
print("Original list:", numbers)

# Remove all occurrences of 2 using filter
numbers = list(filter(lambda x: x != 2, numbers))
print("After filter (removing all 2s):", numbers)
```

## Removing Elements by Index

### Method 1: Using pop() - Removes and returns element at specific index


```python
fruits = ['apple', 'banana', 'cherry', 'date', 'elderberry']
print("Original list:", fruits)

# Remove element at index 2 (cherry)
removed_item = fruits.pop(2)
print("Removed item:", removed_item)
print("List after pop(2):", fruits)

# Remove last element (default behavior of pop())
last_item = fruits.pop()
print("Removed last item:", last_item)
print("Final list:", fruits)
```

### Method 2: Using del statement - Removes element at specific index


```python
colors = ['red', 'green', 'blue', 'yellow', 'purple']
print("Original list:", colors)

# Remove element at index 1 (green)
del colors[1]
print("After del colors[1]:", colors)

# Remove multiple elements using slicing
colors = ['red', 'green', 'blue', 'yellow', 'purple']
print("\nOriginal list:", colors)
del colors[1:3]  # Remove elements from index 1 to 2 (green and blue)
print("After del colors[1:3]:", colors)
```

### Method 3: Removing multiple elements by indices


```python
# Remove multiple elements by specific indices (not consecutive)
numbers = [10, 20, 30, 40, 50, 60, 70]
print("Original list:", numbers)

# Remove elements at indices 1, 3, and 5 (20, 40, 60)
# Important: Remove from highest index to lowest to avoid index shifting
indices_to_remove = [5, 3, 1]  # Remove in reverse order
for index in indices_to_remove:
    numbers.pop(index)

print("After removing indices [1, 3, 5]:", numbers)
```

### Alternative method using list comprehension with enumerate


```python
# Remove multiple elements by indices using list comprehension
numbers = [10, 20, 30, 40, 50, 60, 70]
print("Original list:", numbers)

indices_to_remove = {1, 3, 5}  # Using set for faster lookup
numbers = [item for i, item in enumerate(numbers) if i not in indices_to_remove]
print("After removing indices [1, 3, 5]:", numbers)
```

## Summary of Removal Methods

### Removing All Occurrences:
- **List Comprehension**: `[x for x in list if x != element]` - Most Pythonic
- **filter()**: `list(filter(lambda x: x != element, list))` - Functional approach

### Removing by Index:
- **pop(index)**: Removes and returns element at index
- **del list[index]**: Removes element at index (doesn't return)
- **del list[start:end]**: Removes slice of elements

### Important Notes:
- `remove()` only removes the FIRST occurrence
- When removing multiple indices, do it in reverse order to avoid index shifting
- `pop()` returns the removed element, `del` doesn't
- Always handle IndexError when working with indices


## Combining Two Lists

There are several ways to combine (concatenate) two or more lists in Python.


```python
### Method 1: Using + operator (Concatenation)
# Creates a new list without modifying the original lists

list1 = [1, 2, 3]
list2 = [4, 5, 6]

print("List 1:", list1)
print("List 2:", list2)

combined = list1 + list2
print("Combined list:", combined)
print("List 1 (unchanged):", list1)
print("List 2 (unchanged):", list2)
```

```python
### Method 3: Using * operator (Unpacking)
# Modern Python way using unpacking operator

list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8, 9]

print("List 1:", list1)
print("List 2:", list2)
print("List 3:", list3)

# Combine two lists
combined = [*list1, *list2]
print("\nCombined (list1 + list2):", combined)

# Combine three or more lists
combined_all = [*list1, *list2, *list3]
print("Combined (all three lists):", combined_all)
```

```python
### Method 4: Using list() constructor with + 
# Another way to create a new combined list

list1 = ['apple', 'banana']
list2 = ['cherry', 'date']

combined = list(list1 + list2)
print("Combined list:", combined)
```

```python
### Method 5: Using itertools.chain()
# Efficient for combining multiple lists

from itertools import chain

list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8, 9]

print("List 1:", list1)
print("List 2:", list2)
print("List 3:", list3)

# Combine multiple lists efficiently
print(chain(list1, list2, list3))
combined = list(chain(list1, list2, list3))

print("\nCombined using chain():", combined)
```

```python
### Method 6: Using += operator
# Similar to extend(), modifies the first list

list1 = [1, 2, 3]
list2 = [4, 5, 6]

print("Before +=:")
print("List 1:", list1)
print("List 2:", list2)

list1 += list2  # Same as list1 = list1 + list2, but modifies list1 in place

print("\nAfter list1 += list2:")
print("List 1 (modified):", list1)
print("List 2 (unchanged):", list2)
```
