#### Magic Methods
Magic methods in Python, also known as dunder methods (double underscore methods), are special methods that start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more.

##### Magic Methods
Magic methods are predefined methods in Python that you can override to change the behavior of your objects. Some common magic methods include:
```
__init__: Initializes a new instance of a class.
__str__: Returns a string representation of an object.
__repr__: Returns an official string representation of an object.
__len__: Returns the length of an object.
__getitem__: Gets an item from a container.
__setitem__: Sets an item in a container.
```

````python
class ShoppingCart:
    def __init__(self):
        self.items = {}
    
    # Get item quantity: cart["apple"]
    def __getitem__(self, item_name):
        return self.items.get(item_name, 0)
    
    # Set item quantity: cart["apple"] = 5
    def __setitem__(self, item_name, quantity):
        if quantity > 0:
            self.items[item_name] = quantity
            print(f"Added {quantity} {item_name}(s) to cart")
        else:
            if item_name in self.items:
                del self.items[item_name]
                print(f"Removed {item_name} from cart")
    
    def __delitem__(self, item_name):
        if item_name in self.items:
            del self.items[item_name]
            print(f"Deleted {item_name} from cart")
    
    # String representation
    def __str__(self):
        return f"Cart: {self.items}"

# Usage
cart = ShoppingCart()

# Add items using bracket notation
cart["apple"] = 5       # Added 5 apple(s) to cart
cart["banana"] = 3      # Added 3 banana(s) to cart
cart["orange"] = 2      # Added 2 orange(s) to cart

print(cart)  # Cart: {'apple': 5, 'banana': 3, 'orange': 2}

# Get item quantity
print(f"Apples: {cart['apple']}")    # Apples: 5
print(f"Grapes: {cart['grapes']}")   # Grapes: 0 (not in cart)

# Update quantity
cart["apple"] = 10  # Added 10 apple(s) to cart

# Remove item
del cart["banana"]  # Deleted banana from cart

print(cart)  # Cart: {'apple': 10, 'orange': 2}
````
